### Standard Dataset Example: CIFAR10

A lo largo de tu trabajo con PyTorch, has estado inicializando datasets integrados de PyTorch desde `torchvision` utilizando código similar a este: 

```Python
cifar_dataset = datasets.CIFAR10(root=root_dir, 
                                 train=True, 
                                 download=True,
                                )
```

Pasas algunos parámetros y todo simplemente funciona.

Pero, ¿qué está haciendo exactamente cada uno de estos parámetros iniciales o "flags"? Comprenderlos es el primer paso para dominar cómo manejas los datos en PyTorch. Desglosemos este ejemplo común.

* `root`: Un string que especifica la ruta del directorio donde se almacenará el dataset. Si los datos ya existen en esta ruta, PyTorch los utilizará.

* `train`: Un booleano que selecciona la división (split) de los datos. Para `CIFAR-10`, `True` carga las 50,000 imágenes de entrenamiento, mientras que `False` carga las 10,000 imágenes de prueba.

* `download`: Un booleano que, si es `True`, descarga el dataset de internet si no se encuentra en tu directorio `root`. Si es `False`, ocurrirá un error, **pero solo si el dataset no está ya presente** en la ruta `root`.

Veamos esto en acción.

* Ejecuta la celda de código a continuación para inicializar el dataset [CIFAR10](https://docs.pytorch.org/vision/0.9/datasets.html#torchvision.datasets.CIFAR10).
* Utilizarás el split de entrenamiento configurando `train=True` y permitirás que los datos se descarguen configurando `download=True`.

````python
# Initialize the CIFAR-10 training dataset
cifar_dataset = datasets.CIFAR10(
    root=root_dir,      # Path to the directory where the data is/will be stored
    train=True,         # Specify that you want the training split of the dataset
    download=True       # Download the data if it's not found in the root directory
)
````

Para una descarga simple utilizando las etiquetas (flags) que configuraste, cada elemento en el dataset es una tupla que contiene dos cosas:

* Una **imagen** en formato PIL.
* Un **entero** que representa la etiqueta (label) de esa imagen.

````python
# Get the first sample (at index 0), which is a (image, label) tuple
image, label = cifar_dataset[0]

print(f"Image Type:        {type(image)}")
# Since `image` a PIL Image object, its dimensions are accessed using the .size attribute.
print(f"Image Dimensions:  {image.size}")
print(f"Label Type:        {type(label)}")
````

#### Preparing the Data with Transformations

Como ya sabes, los modelos de PyTorch requieren que los datos de entrada sean **tensors**, no imágenes PIL. También necesitas transformaciones para normalizar los datos y aumentarlos, lo que ayuda a entrenar un modelo robusto.

Después de haber descargado un dataset y quizás haber realizado algún análisis inicial sobre él, puedes prepararlo para tu modelo. Esto se hace asignando un pipeline de transformación directamente al atributo `.transform` del dataset. Esto modifica el dataset in-situ (in-place), de modo que, a partir de ese momento, cada vez que accedas a un elemento, la transformación se aplicará automáticamente.

````python
# Define a transformations pipeline
cifar_transformation = transforms.Compose([
    transforms.ToTensor(),
    # The mean and std values are standard for the CIFAR-10 dataset
    transforms.Normalize(mean=(0.4914, 0.4822, 0.4465),
                         std=(0.2023, 0.1994, 0.2010)
                        )
])

# Assign the entire transformation pipeline to the dataset's .transform attribute
cifar_dataset.transform = cifar_transformation
````

* Observa cómo la transformación se aplica automáticamente.
    * La `image` es ahora un tensor de PyTorch con la forma adecuada, pero la `label` sigue siendo un entero, ya que el pipeline de transformación solo afectó a la `image`.

````python
# Access the first item again
image, label = cifar_dataset[0]

print(f"Image Type:                   {type(image)}")
# Since the `image` is now a PyTorch Tensor, its dimensions are accessed using the .shape attribute.
print(f"Image Shape After Transform:  {image.shape}")
print(f"Label Type:                   {type(label)}")
````

* Usar un DataLoader para tomar un pequeño batch de tus imágenes transformadas y mostrarlas en una cuadrícula.

````python
# Define the datalaoder
cifar_dataloader = data.DataLoader(cifar_dataset, batch_size=8, shuffle=True)
````

````python
helper_utils.display_images(cifar_dataloader)
````

#### El Enfoque Directo: Usando el Parámetro 'transform'

El método que acabas de ver, asignar a `.transform` después de la carga, es potente y flexible, especialmente cuando estás explorando un nuevo dataset.

Sin embargo, en la mayoría de los proyectos del mundo real, ya conoces qué transformaciones necesitas desde el principio. Para esto, existe un enfoque más directo y común. Puedes pasar tu pipeline de transformación directamente al inicializador del dataset utilizando el parámetro `transform`.



```python
cifar_dataset = datasets.CIFAR10(...
                                 transform=cifar_transformation
                                )
```
Esto le indica al dataset que aplique estas transformaciones desde el momento en que se crea, todo en un solo paso limpio.

* Ejecuta la celda de código a continuación para inicializar el dataset `CIFAR10`.

````python
cifar_dataset = datasets.CIFAR10(root=root_dir, 
                                 train=False, 
                                 download=True,
                                 transform=cifar_transformation
                                )
````

### Dataset con Ejemplo de Parámetros Especiales: EMNIST

Si bien muchos datasets comparten parámetros de inicialización similares, algunos tienen requisitos únicos. El dataset EMNIST es uno de esos ejemplos. Echemos un vistazo.

* Primero, define el pipeline de transformación para el dataset `EMNIST`.
    * Dado que las imágenes originales en este dataset no están orientadas correctamente para una visualización fácil, las **rotarás** y las **voltearás verticalmente**.
    * Para la normalización, usarás `transforms.Normalize((0.5,), (0.5,))`. Esto se explicará en breve.

````python
# Definir el pipeline de transformación
emnist_transformation = transforms.Compose([
    # Rotación de 90 grados, rota aleatoriamente entre +90 grados y +90 grados
    transforms.RandomRotation(degrees=(90, 90)),
    # p=1.0 garantiza el volteo vertical (vertical flip)
    transforms.RandomVerticalFlip(p=1.0),
    transforms.ToTensor(),
    # Normaliza el tensor, reescalando los píxeles de [0, 1] a [-1, 1]
    transforms.Normalize((0.5,), (0.5,)) # La media (mean) y desviación estándar (std) deben estar en una tupla
])
````

A diferencia de CIFAR-10, la clase del dataset `EMNIST` requiere un parámetro adicional: `split`. Esto se debe a que el dataset EMNIST es en realidad **una colección de seis datasets diferentes**, y debes especificar cuál deseas cargar.

```python
emnist_digits_dataset = datasets.EMNIST(root=root_dir,
                                        split= ...,  # Need to specify a `split`
                                        train=True,
                                        download=True,
                                        transform=emnist_transformation
                                       )
```

Los splits disponibles son: `byclass`, `bymerge`, `balanced`, `letters`, `digits` y `mnist`.

Debido a que cada split contiene datos diferentes, no existe una única forma estándar de normalizar toda la colección EMNIST.

Para el split `digits` que utilizarás, `transforms.Normalize((0.5,), (0.5,))` es una elección estándar y efectiva. Los valores deben estar en una tupla porque la función espera una secuencia de medias y desviaciones estándar para cada canal de la imagen.

* Ejecuta la celda de código a continuación para inicializar el dataset [EMNIST](https://docs.pytorch.org/vision/0.9/datasets.html#torchvision.datasets.EMNIST).
* Especificarás el split `digits` para cargar solo las imágenes de dígitos escritos a mano.

````python
emnist_digits_dataset = datasets.EMNIST(root=root_dir,
                                        split='digits',  # Specify the 'digits' split
                                        train=False,
                                        download=True,
                                        transform=emnist_transformation
                                       )
````

* Usa un `DataLoader` para tomar un pequeño batch de tus imágenes `EMNIST` cargadas y mostrarlas en una cuadrícula.

````python
emnist_digits_dataloader = data.DataLoader(emnist_digits_dataset, batch_size=8, shuffle=True)
````

````python
helper_utils.display_images(emnist_digits_dataloader)
````

## Custom and Specialized Datasets

Más allá de los datasets predefinidos, TorchVision proporciona herramientas potentes para trabajar con tus propias imágenes y generar datos de prueba.

### Loading Custom Images: ImageFolder

`ImageFolder` es un cargador de datasets genérico que te permite usar tus propias imágenes sin tener que escribir una clase `Dataset` personalizada. Es increíblemente útil, pero requiere que organices tus imágenes en una estructura de carpetas específica.

El requisito principal para `ImageFolder` es que tu directorio raíz debe contener un subdirectorio para cada clase. `ImageFolder` utiliza automáticamente el nombre de cada subdirectorio como la etiqueta (label) de la clase.



Para este ejemplo, utilizarás la siguiente estructura de directorios:

```
./tiny_fruit_and_vegetable/
├── Apple__Healthy/
│   ├── FreshApple (2).jpg
│   └── ...
├── Guava__Healthy/
│   ├── FreshGuava (179).jpg
│   └── ...
├── Mango__Healthy/
│   ├── 22.jpg
│   └── ...
├── Pomegranate__Healthy/
│   ├── FreshPomegranate (3).jpg
│   └── ...
└── Strawberry__Healthy/
    ├── 48.jpg
    └── ...
```

Cuando apuntes `ImageFolder` al directorio `./tiny_fruit_and_vegetable`, este:

* Encontrará automáticamente **cinco clases** basándose en los nombres de los subdirectorios: `Apple__Healthy`, `Guava__Healthy`, `Mango__Healthy`, `Pomegranate__Healthy` y `Strawberry__Healthy`.
* Asignará una etiqueta entera a cada clase (por ejemplo, `Apple__Healthy` --> `0`, `Guava__Healthy` --> `1`, y así sucesivamente).
* Asumirá que todas las imágenes dentro de la carpeta `Apple__Healthy/` pertenecen a la clase `0`, todas las imágenes en `Guava__Healthy/` pertenecen a la clase `1`, etc.

**NOTA:** puedes explorar la carpeta en la barra lateral izquierda.

* Define the path for the `tiny_fruit_and_vegetable` directory of the dataset. `ImageFolder` will use this path to find the images.

````python
root_dir = './tiny_fruit_and_vegetable'
````

Al igual que los datasets integrados, `ImageFolder` te permite pasar un pipeline de transformación para que se aplique a tus imágenes a medida que se cargan. Para cualquier tarea de deep learning, este pipeline debe, **como mínimo, convertir las imágenes en tensors**. Además, para permitir el procesamiento por lotes (batch processing) con un `DataLoader`, también es esencial **redimensionar las imágenes a un tamaño uniforme**.

Para este ejemplo, también normalizarás las imágenes. Si bien la mejor práctica es utilizar la media y la desviación estándar *reales* de tu dataset específico para un rendimiento óptimo, usar `0.5` para ambos es un punto de partida común y efectivo que escala los valores de tus píxeles al rango `[-1, 1]`.

````python
# Define a transformation pipeline
image_transformation = transforms.Compose([
    transforms.Resize((100, 100)),
    transforms.ToTensor(),
    transforms.Normalize( 
        mean=[0.5, 0.5, 0.5],
        std=[0.5, 0.5, 0.5]
    )
])
````

* Ejecuta la celda de código a continuación para inicializar el dataset [ImageFolder](https://docs.pytorch.org/vision/main/generated/torchvision.datasets.ImageFolder.html).
* Pasarás los siguientes argumentos:
    * `root`: La ruta al directorio raíz que contiene las subcarpetas de tus clases.
    * `transform`: Un argumento **opcional** para el pipeline de transformación que se aplicará. Aunque puedes pasar `None`, en este caso, utilizarás el pipeline que definiste anteriormente.

````python
fruit_dataset = datasets.ImageFolder(root=root_dir,
                                     transform=image_transformation
                                    )
````

> #### Nota Adicional:
>
> Como has visto, cada dataset es único. Es posible que hayas notado que, a diferencia de `CIFAR10` o `EMNIST`, aquí no pasaste un argumento `train=True` o `train=False`. Esto se debe a que `ImageFolder` funciona con tu propio directorio personalizado de imágenes, el cual no tiene una división predefinida.
>
> Entonces, ¿cómo se crean conjuntos de entrenamiento y de validación/prueba al usar `ImageFolder`? Aunque hay varias formas de abordar esto, dos estrategias son particularmente comunes.
>
> **Estrategia 1: Dividir el Objeto del Dataset**
>
> La primera estrategia consiste en cargar todo el dataset desde tu carpeta raíz en un único objeto `ImageFolder`. Luego, usarías una función de PyTorch, como `torch.utils.data.random_split`, para dividir programáticamente ese único objeto de dataset en datasets más pequeños de entrenamiento y prueba. Este enfoque es muy conveniente si planeas aplicar exactamente las mismas transformaciones a ambas divisiones.
>
> **Estrategia 2: Dividir los Directorios Previamente**
>
> La segunda estrategia común es organizar tu sistema de archivos en subdirectorios `train` y `test` incluso antes de ejecutar tu código. Cada uno de estos directorios contendría las subcarpetas de clases en su interior. Luego, crearías dos datasets `ImageFolder` separados, uno apuntando al directorio "train" y el otro al directorio "test". Este método es ideal cuando necesitas aplicar diferentes pipelines de transformación a tus datos de entrenamiento y prueba, como incluir aumentos de datos (data augmentation) solo para el conjunto de entrenamiento.

* Usa un `DataLoader` para tomar un pequeño batch de tus imágenes cargadas con `ImageFolder` y mostrarlas en una cuadrícula.

````python
fruit_dataloader = data.DataLoader(fruit_dataset, batch_size=8, shuffle=True)
````

````python
helper_utils.display_images(fruit_dataloader)
````

### Generando Datos Sintéticos: FakeData

`FakeData` es un dataset de `torchvision` muy práctico que genera imágenes y etiquetas aleatorias sobre la marcha. Es perfecto para probar rápidamente un bucle de entrenamiento, un pipeline de carga de datos o una arquitectura de modelo sin tener que esperar a que se descargue un dataset real. Puedes especificar propiedades clave como el número de imágenes, sus dimensiones y el número de clases.

No entra en la categoría de "datasets personalizados" en el sentido tradicional, ya que no estás proporcionando los datos, sino que es una herramienta especializada para el desarrollo y la depuración.

* Ejecuta la celda de código a continuación para inicializar el dataset [FakeData](https://docs.pytorch.org/vision/main/generated/torchvision.datasets.FakeData.html#torchvision.datasets.FakeData).
* Pasarás los siguientes argumentos:
    * `size`: El número total de muestras falsas que deseas en el dataset.
    * `image_size`: Una tupla que define las dimensiones de las imágenes generadas en el formato `(Canales, Altura, Ancho)`.
    * `num_classes`: El número de clases posibles para las etiquetas.
    * `transform`: Un pipeline de transformación **opcional** para aplicar a las imágenes a medida que se generan.

````python
# Define a transformation pipeline
fake_data_transform = transforms.Compose([
    transforms.ToTensor()
])

# Initialize the FakeData dataset
fake_dataset = datasets.FakeData(
    size=1000,                    # Total number of fake images
    image_size=(3, 32, 32),       # (Channels, Height, Width)
    num_classes=10,               # Number of possible classes
    transform=fake_data_transform # Apply the transformation
)
````

* Usa un `DataLoader` para tomar un pequeño batch de tus imágenes generadas con `FakeData` y mostrarlas en una cuadrícula.

````python
fake_dataloader = data.DataLoader(fake_dataset, batch_size=8, shuffle=True)
````

````python
helper_utils.display_images(fake_dataloader)
````
